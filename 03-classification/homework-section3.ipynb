```python
df = pd.read_csv(
    "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/course_lead_scoring_2026.csv"
)

print(df.shape)
```

```python
categorical = [
    'lead_source', 'industry',
    'employment_status', 'location'
]

numerical = [
    'annual_income', 'number_of_courses_viewed',
    'interaction_count', 'lead_score'
]

df[categorical] = df[categorical].fillna('NA')
df[numerical] = df[numerical].fillna(0.0)

print("Q1 — Industry mode:")
print(df['industry'].mode()[0])
print(df['industry'].value_counts())
```

```python
correlations = df[numerical].corr()

print("Q2 — Correlation matrix:")
display(correlations)

pairs = [
    ('interaction_count', 'lead_score'),
    ('number_of_courses_viewed', 'lead_score'),
    ('number_of_courses_viewed', 'interaction_count'),
    ('annual_income', 'interaction_count')
]

for a, b in pairs:
    print(a, "and", b, round(correlations.loc[a, b], 6))

print("Highest:",
      max(pairs, key=lambda p: correlations.loc[p[0], p[1]]))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mutual_info_score

df_full_train, df_test = train_test_split(
    df, test_size=0.2, random_state=42
)

df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=42
)

y_train = df_train['converted'].values
y_val = df_val['converted'].values

X_train_df = df_train.drop(columns=['converted'])
X_val_df = df_val.drop(columns=['converted'])

print("Q3 — Mutual information:")

mi = {}

for col in categorical:
    mi[col] = mutual_info_score(
        X_train_df[col], y_train
    )

print(pd.Series(mi).sort_values(ascending=False).round(6))
```

```python
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

features = categorical + numerical

def fit_and_score(features, C=1.0):
    dv = DictVectorizer(sparse=False)

    train_dict = X_train_df[features].to_dict(orient='records')
    val_dict = X_val_df[features].to_dict(orient='records')

    X_train = dv.fit_transform(train_dict)
    X_val = dv.transform(val_dict)

    model = LogisticRegression(
        solver='liblinear',
        C=C,
        max_iter=1000,
        random_state=42
    )

    model.fit(X_train, y_train)

    predictions = model.predict(X_val)

    return accuracy_score(y_val, predictions)

full_accuracy = fit_and_score(features)

print("Q4 — Accuracy:", full_accuracy)
print("Rounded:", round(full_accuracy, 2))
```

```python
candidates = [
    'lead_source',
    'number_of_courses_viewed',
    'interaction_count'
]

results = []

for removed in candidates:
    remaining = [f for f in features if f != removed]
    reduced_accuracy = fit_and_score(remaining)

    results.append({
        'Removed feature': removed,
        'Reduced accuracy': reduced_accuracy,
        'Signed difference': full_accuracy - reduced_accuracy
    })

results_df = pd.DataFrame(results)

display(results_df.sort_values('Signed difference'))

print("Q5 — Smallest signed difference:",
      results_df.loc[
          results_df['Signed difference'].idxmin(),
          'Removed feature'
      ])
```

```python
c_values = [0.000001, 0.00001, 0.0001, 0.001]

scores = []

for C in c_values:
    score = fit_and_score(features, C=C)
    scores.append((C, round(score, 3)))

results_c = pd.DataFrame(
    scores, columns=['C', 'Accuracy']
)

display(results_c)

best_accuracy = results_c['Accuracy'].max()
best_C = results_c.loc[
    results_c['Accuracy'] == best_accuracy, 'C'
].min()

print("Q6 — Best C:", best_C)
```
