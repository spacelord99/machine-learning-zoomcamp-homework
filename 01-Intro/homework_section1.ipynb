```python
median_before = df_hw['horsepower'].median()
median_before
```

```python
mean_hp = df_hw['horsepower'].mean()
mean_hp
```

```python
df_filled = df_hw.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(mean_hp)
```

```python
median_after = df_filled['horsepower'].median()
median_after
```

```python
median_before, median_after
```

```python
mode_hp = df_hw['horsepower'].mode()[0]
mode_hp
```

```python
median_before = df_hw['horsepower'].median()

df_mode = df_hw.copy()
df_mode['horsepower'] = df_mode['horsepower'].fillna(mode_hp)

median_after = df_mode['horsepower'].median()

median_before, median_after
```

```python
import numpy as np
```

```python
asia = df_hw[df_hw['origin'] == 'Asia']
```

```python
asia.head()
```

```python
asia = asia[['vehicle_weight', 'model_year']]
```

```python
asia = asia.head(7)
asia
```

```python
X = asia.values
X
```

```python
X.shape
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
y
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```
